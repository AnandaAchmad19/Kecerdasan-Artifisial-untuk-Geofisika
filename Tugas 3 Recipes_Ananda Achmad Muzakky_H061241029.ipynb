## 2.1 Recipe Overview
Dalam berbagai fenomena kebumian, suatu gaya dorong atau pemicu fisik (*forcing mechanism*) tidak serta merta menghasilkan respons seketika pada sistem bawah permukaan. Fenomena ini dikenal sebagai keterlambatan fisis (*time lag* atau *phase delay*). Contoh paling nyata dijumpai dalam hidrogeofisika: ketika terjadi hujan lebat di permukaan tanah, air memerlukan waktu untuk berinfiltrasi melalui zona tak jenuh (*unsaturated vadose zone*) sebelum akhirnya mencapai muka air tanah (*water table*) dan menyebabkan kenaikan elevasi hidraulik akuifer.

### Masalah yang Diselesaikan
Menentukan durasi jeda waktu ($	au$) yang paling optimal antara dua deret waktu geofisika, serta mengukur derajat keterikatan linier pada setiap pergeseran waktu tersebut secara objektif.

### Tujuan
1. Menghitung fungsi korelasi silang ternormalisasi (*Normalized Cross-Correlation Function / CCF*) untuk rentang pergeseran waktu lag $[-	au_{max}, +	au_{max}]$.
2. Mengidentifikasi nilai lag optimal ($	au^*$) yang menghasilkan korelasi tertinggi ($r_{max}$).
3. Menilai signifikansi statistik dari lag tersebut untuk membedakan respons geofisika sejati dari korelasi acak atau semu (*spurious correlation*).

### Kegunaan dalam Geofisika
- **Hidrogeofisika**: Menentukan waktu infiltrasi dan kecepatan peresapan air tanah dari data stasiun meteorologi dan piezometer.
- **Vulkanologi**: Mengkorelasikan laju pelepasan gas SO₂ atau inflasi deformasi tiltmeter dengan intensitas gempa tremor vulkanik menjelang erupsi.
- **Seismologi Lingkungan**: Analisis korelasi silang *ambient seismic noise* antar-stasiun stasiun seismometer untuk merekonstruksi fungsi Green (*empirical Green's functions*).
- **Geodinamika & Oseanografi**: Menghubungkan anomali angin laut atau pasang surut dengan penurunan tanah terukur GPS (*subsidence*).

### Hasil yang Diharapkan
Array nilai lag terkalibrasi (dalam satuan fisik seperti hari atau jam), koefisien korelasi Pearson ternormalisasi $r \in [-1, 1]$, nilai lag puncak terbaik ($	au^*$), batas signifikansi statistik (95% Confidence Interval), serta visualisasi interaktif deret waktu dan correlogram.


## 2.2 When to Use

### Kondisi Sesuai Penggunaan
- Terdapat hubungan kausalitas terarah (*cause-and-effect relationship*) antara sinyal input pemicu $x(t)$ dan respons sistem $y(t)$.
- Kedua deret waktu memiliki interval perekaman (*sampling interval* $\Delta t$) yang konstan, seragam, dan identik.
- Panjang data deret waktu cukup memadai ($N \gg 	au_{max}$) sehingga estimasi kovariansi pada pergeseran lag ekstrem tetap memiliki derajat kebebasan statistik yang valid.

### Asumsi yang Diperlukan
1. **Kuasi-Stasioneritas (*Stationarity*)**: Deret waktu harus bebas dari tren jangka panjang (*secular trend*) yang drastis; rata-rata dan varians tidak berubah secara eksponensial selama jendela observasi.
2. **Keterikatan Linier (*Linearity*)**: Hubungan antara pemicu dan respons diasumsikan dapat didekati dengan sistem linier invarian waktu (*Linear Time-Invariant / LTI*).
3. **Sampling Uniform**: Tidak terdapat celah data (*missing data gaps*) yang signifikan pada interval penjelajahan lag.

### Keterbatasan
- **Korelasi Bukan Kausalitas Murni**: Korelasi silang yang tinggi dapat timbul secara kebetulan akibat variabel ketiga yang tidak teramati (*confounding variable*).
- **Rentan terhadap Autokorelasi Tinggi (*Red Noise*)**: Sinyal geofisika dengan persistensi internal tinggi (misalnya deret muka air tanah) dapat menunjukkan kurva korelasi lebar dan semu jika tidak dilakukan *detrending* atau *pre-whitening*.
- **Sensitif terhadap Outlier**: Lonjakan ekstrem instrumen dapat mendistorsi nilai puncak korelasi Pearson.

## 2.4 Konsep dan Algoritma

### 1. Definisi Matematis Korelasi Silang (*Cross-Correlation*)
Korelasi silang antara deret waktu pemicu $x(t)$ dan deret waktu respons $y(t)$ pada pergeseran lag waktu $\tau$ dirumuskan sebagai:

$$r_{xy}(\tau) = \frac{\sum_{t} \big(x(t) - \bar{x}\big) \big(y(t + \tau) - \bar{y}\big)}{\sqrt{\sum_{t} \big(x(t) - \bar{x}\big)^2} \cdot \sqrt{\sum_{t} \big(y(t) - \bar{y}\big)^2}}$$

Di mana:
- $\bar{x}$ dan $\bar{y}$ adalah nilai rata-rata dari masing-masing deret waktu.
- Pembilang merepresentasikan kovariansi silang (*cross-covariance*) tergeser waktu.
- Penyebut merepresentasikan normalisasi oleh deviasi standar terbobot sampel, menjamin $r_{xy}(\tau) \in [-1.0, +1.0]$.

### 2. Konvensi Lag Positif vs Lag Negatif
Pemberian tanda lag sangat penting untuk menghindari kesalahan interpretasi fisik:
- **Lag Positif ($\tau > 0$)**: Sinyal $y$ digeser mundur relatif terhadap $x$, yang ekuivalen dengan kondisi di mana $x$ mendahului $y$ dalam domain waktu nyata ($x(t)$ menyebabkan respons $y(t + \tau)$). Dalam konteks hidrogeologi, puncak korelasi pada $\tau > 0$ menandakan curah hujan memicu kenaikan air tanah beberapa hari kemudian.
- **Lag Negatif ($\tau < 0$)**: Sinyal $y$ mendahului $x$. Jika puncak berada di lag negatif, hipotesis bahwa $x$ memicu $y$ tidak valid secara fisis.
- **Lag Nol ($\tau = 0$)**: Respons terjadi secara instan tanpa penundaan terukur.

### 3. Detrending dan Standardisasi
Sinyal mentah geofisika sering kali memiliki tren sekuler (misalnya tren penurunan muka air tanah regional akibat eksploitasi sumur bor jangka panjang). Jika tren linier $y_{trend}(t) = at + b$ tidak dibuang, hasil korelasi silang akan didominasi oleh kemiripan tren linier, bukan dinamika fluktuasi hujan. 

Proses detrending dilakukan dengan metode kuadrat terkecil (*ordinary least squares*):
$$x_{detrend}(t) = x(t) - (a_x t + b_x)$$
$$y_{detrend}(t) = y(t) - (a_y t + b_y)$$

### 4. Ambang Batas Signifikansi Statistik (Formula Bartlett)
Berdasarkan pendekatan Bartlett untuk deret waktu independen berdistribusi normal, batas selang kepercayaan 95% di sekitar nol dihitung dengan:
$$CI_{95} = \pm \frac{1.96}{\sqrt{N}}$$
Di mana $N$ adalah jumlah titik sampel observasi. Nilai $|r_{xy}(\tau)| > CI_{95}$ mengindikasikan bahwa korelasi tersebut secara statistik berbeda signifikan dari nol pada tingkat kepercayaan 95%.


## 2.5 Implementasi Langkah demi Langkah
Bagian ini memaparkan langkah-langkah komputasi bertahap untuk merekonstruksi dan menganalisis fenomena respon muka air tanah terhadap curah hujan:

1. **Sintesis Data Hujan**: Membangkitkan deret curah hujan harian realistis menggunakan distribusi terputus nol (*zero-inflated Gamma distribution*), mencerminkan pola hari kering dan hari basah.
2. **Sintesis Respon Air Tanah Tertunda**: Memodelkan perambatan air melalui zona tak jenuh dengan fungsi respon impuls eksponensial tertunda (*delayed exponential filter*) dengan *true lag* yang diset tepat sebesar 7 hari, ditambah tren musiman dan derau acak Gaussian.
3. **Pemeriksaan Kualitas Data**: Memvalidasi keseragaman panjang vektor, ketiadaan nilai `NaN`/`inf`, serta ringkasan statistika dasar.
4. **Komputasi Korelasi Silang**: Menghitung koefisien korelasi silang untuk setiap pergeseran waktu.
5. **Identifikasi Lag Puncak**: Menentukan posisi indeks lag dengan koefisien korelasi absolut maksimum.
6. **Visualisasi Komprehensif**: Menampilkan kedua deret waktu dan kurva korelasi silang (*correlogram*).


```python
# =============================================================================
# LANGKAH 1 & 2: SINTESIS DERET WAKTU CURAH HUJAN DAN RESPON MUKA AIR TANAH
# =============================================================================
n_days = 365
days = np.arange(n_days)
dt_days = 1.0  # Sampling 1 hari

# Pembangkitan curah hujan harian (Zero-inflated Gamma process)
prob_rain = 0.28  # Peluang terjadi hujan ~28% hari
rain_occurrence = np.random.rand(n_days) < prob_rain
rainfall_raw = np.zeros(n_days)
# Intensitas hujan saat hari basah (skala gamma: bentuk=2.2, skala=9.0 mm)
rainfall_raw[rain_occurrence] = np.random.gamma(shape=2.2, scale=9.0, size=np.sum(rain_occurrence))

# Respons muka air tanah dengan Delay Fisik Sejati (True Lag = 7 Hari)
TRUE_LAG = 7  # Hari
kernel_length = 30
t_k = np.arange(kernel_length)

# Kernel transfer function: Infiltrasi zona vadose dengan penundaan 7 hari
response_kernel = np.zeros(kernel_length)
response_kernel[TRUE_LAG:] = np.exp(-(t_k[TRUE_LAG:] - TRUE_LAG) / 3.5)
response_kernel /= np.sum(response_kernel)  # Normalisasi luas kernel

# Konvolusi curah hujan dengan fungsi transfer akuifer
gw_recharge = np.convolve(rainfall_raw, response_kernel, mode='full')[:n_days]

# Menambahkan dinamika hidrologi: elevasi dasar (base level) + variasi musiman + noise instrumen
gw_base = 12.0  # Meter di atas datum
seasonal_oscillation = 0.8 * np.sin(2 * np.pi * days / 365.0 - 0.5)
measurement_noise = np.random.normal(loc=0.0, scale=0.08, size=n_days)

# Sinyal elevasi muka air tanah sintetis lengkap (meter)
gw_level_raw = gw_base + seasonal_oscillation + 0.15 * gw_recharge + measurement_noise

# =============================================================================
# LANGKAH 3: PEMERIKSAAN KUALITAS DATA (DATA QUALITY CHECK)
# =============================================================================
assert len(rainfall_raw) == len(gw_level_raw), "Panjang kedua deret waktu harus identik!"
assert not np.isnan(rainfall_raw).any(), "Data curah hujan mengandung nilai NaN!"
assert not np.isnan(gw_level_raw).any(), "Data air tanah mengandung nilai NaN!"

print("=== RINGKASAN PEMERIKSAAN KUALITAS DATA ===")
print(f"Jumlah observasi sampel : {n_days} hari")
print(f"Curah Hujan Maksimum    : {np.max(rainfall_raw):.2f} mm/hari | Rata-rata: {np.mean(rainfall_raw):.2f} mm/hari")
print(f"Muka Air Tanah Min / Max: {np.min(gw_level_raw):.2f} m / {np.max(gw_level_raw):.2f} m | Dev. Std: {np.std(gw_level_raw):.3f} m")
print(f"Parameter True Lag fisis: {TRUE_LAG} hari (ditanamkan dalam kernel respon)\n")

# =============================================================================
# LANGKAH 4 & 5: PERHITUNGAN LAG CORRELATION LANGKAH-DEMI-LANGKAH
# =============================================================================
# Detrending linier sederhana pada muka air tanah
p_fit = np.polyfit(days, gw_level_raw, 1)
gw_detrended = gw_level_raw - np.polyval(p_fit, days)

# Normalisasi z-score
x_norm = (rainfall_raw - np.mean(rainfall_raw)) / np.std(rainfall_raw)
y_norm = (gw_detrended - np.mean(gw_detrended)) / np.std(gw_detrended)

max_search_lag = 25
lags_manual = np.arange(-max_search_lag, max_search_lag + 1)
corr_manual = []

for lag in lags_manual:
    if lag > 0:
        # x mendahului y: iris x[:-lag] dan y[lag:]
        c = np.mean(x_norm[:-lag] * y_norm[lag:])
    elif lag < 0:
        # y mendahului x: iris x[-lag:] dan y[:lag]
        c = np.mean(x_norm[-lag:] * y_norm[:lag])
    else:
        c = np.mean(x_norm * y_norm)
    corr_manual.append(c)

corr_manual = np.array(corr_manual)
idx_max = np.argmax(corr_manual)
estimated_lag_manual = lags_manual[idx_max]
peak_corr_manual = corr_manual[idx_max]

print(f"Hasil Langkah-demi-Langkah:")
print(f"-> Estimasi Lag Terbaik : {estimated_lag_manual} hari")
print(f"-> Koefisien Korelasi   : {peak_corr_manual:.4f}")
print(f"-> Selisih vs True Lag  : {abs(estimated_lag_manual - TRUE_LAG)} hari")

# =============================================================================
# LANGKAH 6: VISUALISASI LANGKAH AWAL
# =============================================================================
fig, axs = plt.subplots(2, 1, figsize=(10, 6), sharex=False, gridspec_kw={'hspace': 0.35})

# Panel 1: Deret Waktu Curah Hujan dan Muka Air Tanah
ax1_twin = axs[0].twinx()
p1 = axs[0].bar(days, rainfall_raw, width=0.8, color='#3498DB', alpha=0.75, label='Curah Hujan (mm/hari)')
p2 = ax1_twin.plot(days, gw_level_raw, color='#E74C3C', lw=1.8, label='Muka Air Tanah (m)')
axs[0].set_ylabel('Curah Hujan (mm)', color='#2980B9', fontweight='bold')
ax1_twin.set_ylabel('Elevasi MAT (m)', color='#C0392B', fontweight='bold')
axs[0].set_xlabel('Waktu (Hari ke-n)')
axs[0].set_title('(A) Data Observasi Deret Waktu Sintetis Curah Hujan & Muka Air Tanah', fontweight='bold')

# Panel 2: Kurva Korelasi Silang
axs[1].plot(lags_manual, corr_manual, color='#2C3E50', lw=2.0, marker='o', ms=4, label='Cross-Correlation')
axs[1].axvline(TRUE_LAG, color='green', ls='--', lw=1.8, label=f'True Lag = {TRUE_LAG} hari')
axs[1].axvline(estimated_lag_manual, color='red', ls=':', lw=2.0, label=f'Peak Lag = {estimated_lag_manual} hari')
axs[1].axhline(0, color='black', lw=0.8, ls='-')
axs[1].axhline(1.96 / np.sqrt(n_days), color='gray', ls='--', label='95% CI Limit')
axs[1].axhline(-1.96 / np.sqrt(n_days), color='gray', ls='--')
axs[1].set_xlabel('Pergeseran Waktu Lag (Hari)')
axs[1].set_ylabel('Koefisien Korelasi $r_{xy}$')
axs[1].set_title('(B) Fungsi Korelasi Silang (CCF) vs Lag Waktu', fontweight='bold')
axs[1].legend(loc='upper right', frameon=True)

plt.show()
```

## 2.6 Fungsi Reusable: `compute_lag_correlation`

Untuk memenuhi standar *engineering code* yang dapat dipakai ulang (*reusable*) pada aneka macam data deret waktu geofisika, algoritma di atas dikemas ke dalam sebuah fungsi Python profesional bernama `compute_lag_correlation`.

### Fitur Kunci Fungsi:
1. **Validasi Parameter Lengkap**: Memeriksa dimensi larik 1D, konsistensi panjang data, ketiadaan nilai invalid (`NaN`, `Inf`), dan kelayakan rentang batas `max_lag`.
2. **Fleksibilitas Satuan Waktu**: Mengintegrasikan parameter `dt` dan `time_unit` sehingga otomatis menghasilkan sumbu waktu lag dalam satuan fisik riil (misal detik, jam, hari, bulan).
3. **Pembersihan Tren Otomatis (*Detrending*)**: Mendukung *ordinary least squares polynomial detrending* untuk membuang tren sekuler non-stasioner.
4. **Keluaran Terstruktur (*Structured Dictionary*)**: Mengembalikan dictionary lengkap yang memuat vektor lag, koefisien korelasi, posisi lag optimal, signifikansi statistik (95% CI), dan metadata analisis.


```python
def compute_lag_correlation(
    x: np.ndarray,
    y: np.ndarray,
    dt: float = 1.0,
    max_lag: int = 30,
    detrend: bool = True,
    normalize: bool = True,
    time_unit: str = "hari"
) -> dict:
    """
    Menghitung fungsi korelasi silang bertingkat (lag cross-correlation) antara dua deret waktu geofisika.

    Fungsi ini dirancang reusable untuk mengidentifikasi jeda waktu (time lag) antara sinyal pemicu
    (driving signal x) dan respons sistem (response signal y), misalnya curah hujan vs muka air tanah,
    tekanan udara vs deformasi, atau gempa vulkanik vs pelepasan gas.

    Parameters
    ----------
    x : np.ndarray
        Array 1D deret waktu sinyal pemicu (input/forcing time series).
    y : np.ndarray
        Array 1D deret waktu sinyal respons (system response time series).
    dt : float, default=1.0
        Interval sampling temporal antar titik data berurutan (dalam satuan `time_unit`).
    max_lag : int, default=30
        Jumlah sampel lag maksimum yang akan diuji secara simetris [-max_lag, +max_lag].
    detrend : bool, default=True
        Jika True, melakukan detrending linier dengan Least Squares sebelum menghitung korelasi.
    normalize : bool, default=True
        Jika True, menormalkan koefisien korelasi ke rentang baku Pearson [-1.0, 1.0].
    time_unit : str, default='hari'
        Label string satuan waktu untuk dokumentasi keluaran dan visualisasi.

    Returns
    -------
    dict
        Dictionary berisi luaran komputasi:
        - 'lags_index'      : np.ndarray indeks lag integer [-max_lag, ..., max_lag].
        - 'lags_time'       : np.ndarray pergeseran waktu dalam satuan fisis (lags_index * dt).
        - 'correlation'     : np.ndarray koefisien korelasi silang pada setiap lag.
        - 'best_lag_index'  : int indeks lag dengan korelasi puncak.
        - 'best_lag_time'   : float nilai lag fisis optimal dengan korelasi puncak.
        - 'max_correlation' : float nilai koefisien korelasi maksimum.
        - 'ci_95'           : float ambang batas signifikansi dua sisi 95% (Bartlett).
        - 'time_unit'       : str label satuan waktu yang digunakan.
        - 'n_samples'       : int jumlah titik observasi.
        - 'metadata'        : dict informasi tambahan pemrosesan data.

    Raises
    ------
    ValueError
        Jika panjang x dan y berbeda, data kosong, atau max_lag tidak valid.
    TypeError
        Jika input bukan array numerik.
    """
    # 1. Validasi Tipe dan Bentuk Array Input
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)

    if x.ndim != 1 or y.ndim != 1:
        raise ValueError(f"Dimensi x ({x.ndim}D) dan y ({y.ndim}D) harus 1D array tunggal!")
    
    n_samples = len(x)
    if n_samples != len(y):
        raise ValueError(f"Panjang vektor x ({n_samples}) dan y ({len(y)}) harus identik!")
    
    if n_samples < 5:
        raise ValueError("Jumlah sampel data terlalu sedikit (minimal 5 data) untuk korelasi!")
    
    if np.isnan(x).any() or np.isnan(y).any():
        raise ValueError("Deret waktu input mengandung nilai NaN. Harap lakukan imputasi terlebih dahulu!")
    
    if np.isinf(x).any() or np.isinf(y).any():
        raise ValueError("Deret waktu input mengandung nilai Infinite (Inf)!")

    # 2. Validasi Parameter Lag
    if not isinstance(max_lag, (int, np.integer)):
        raise TypeError("Parameter max_lag harus bertipe integer!")
    
    if max_lag <= 0 or max_lag >= n_samples:
        raise ValueError(f"max_lag ({max_lag}) harus bernilai positif dan lebih kecil dari panjang data ({n_samples})!")

    if dt <= 0:
        raise ValueError(f"Interval sampling dt ({dt}) harus bernilai positif!")

    # 3. Pra-pemrosesan Sinyal: Detrending
    x_proc = x.copy()
    y_proc = y.copy()

    if detrend:
        time_vec = np.arange(n_samples)
        p_x = np.polyfit(time_vec, x_proc, deg=1)
        p_y = np.polyfit(time_vec, y_proc, deg=1)
        x_proc = x_proc - np.polyval(p_x, time_vec)
        y_proc = y_proc - np.polyval(p_y, time_vec)

    # 4. Standardisasi Sinyal (Mean-Centering dan Penskalaan Deviasi Standar)
    mean_x, std_x = np.mean(x_proc), np.std(x_proc)
    mean_y, std_y = np.mean(y_proc), np.std(y_proc)

    if std_x == 0 or std_y == 0:
        raise ValueError("Salah satu sinyal memiliki varians nol (konstan), korelasi tidak terdefinisi!")

    x_norm = (x_proc - mean_x) / std_x
    y_norm = (y_proc - mean_y) / std_y

    # 5. Komputasi Korelasi Silang pada Rentang Lag
    lag_indices = np.arange(-max_lag, max_lag + 1)
    correlations = np.zeros(len(lag_indices), dtype=np.float64)

    for i, lag in enumerate(lag_indices):
        if lag > 0:
            # Sinyal x mendahului sinyal y (y merespons setelah jeda lag)
            val = np.mean(x_norm[:-lag] * y_norm[lag:])
        elif lag < 0:
            # Sinyal y mendahului sinyal x
            val = np.mean(x_norm[-lag:] * y_norm[:lag])
        else:
            val = np.mean(x_norm * y_norm)
            
        correlations[i] = val if normalize else val * (std_x * std_y)

    # 6. Identifikasi Puncak Lag Optimal
    idx_peak = np.argmax(np.abs(correlations))
    best_lag_idx = int(lag_indices[idx_peak])
    best_lag_time = float(best_lag_idx * dt)
    max_corr_val = float(correlations[idx_peak])

    # 7. Ambang Batas Signifikansi Dua Sisi 95% (Bartlett Approximation)
    ci_95 = 1.96 / np.sqrt(n_samples)

    return {
        'lags_index': lag_indices,
        'lags_time': lag_indices * dt,
        'correlation': correlations,
        'best_lag_index': best_lag_idx,
        'best_lag_time': best_lag_time,
        'max_correlation': max_corr_val,
        'ci_95': float(ci_95),
        'time_unit': time_unit,
        'n_samples': n_samples,
        'metadata': {
            'detrended': detrend,
            'normalized': normalize,
            'sampling_dt': dt,
            'std_x': float(std_x),
            'std_y': float(std_y)
        }
    }

print("[OK] Fungsi reusable 'compute_lag_correlation' berhasil didefinisikan dan siap digunakan.")
```

## 2.8 Validation and Troubleshooting

Dalam praktiknya, analisis korelasi deret waktu geofisika sering dihadapkan pada jebakan statistik berupa **korelasi semu (*spurious correlation*)**. Korelasi semu dapat terjadi apabila salah satu atau kedua deret waktu memiliki sifat non-stasioner, tren jangka panjang yang dominan, atau autokorelasi internal yang sangat kuat (*red noise persistence*).

### Pengujian Validasi Terkontrol
Di bawah ini kita melakukan dua pengujian diagnostik kritis:
1. **Uji Validasi Sintetik murni (*Pure Shift Calibration*)**: Memvalidasi fungsi terhadap dua sinyal dengan pergeseran waktu eksak yang telah diketahui (misalnya $	au = 12$ hari) tanpa derau, untuk membuktikan akurasi komputasi algoritma.
2. **Eksperimen Korelasi Semu (*Spurious Correlation Trap*)**: Membandingkan hasil korelasi antara data mentah dengan tren musiman kuat vs data yang telah diproses dengan *detrending*.


```python
# =============================================================================
# VALIDASI 1: PURE SHIFT TEST (UJI PRESIFIKASI LAG TERKONTROL)
# =============================================================================
t_val = np.arange(200)
shift_known = 12  # Hari
sig_x = np.sin(2 * np.pi * t_val / 25.0) + 0.5 * np.sin(2 * np.pi * t_val / 10.0)
sig_y = np.roll(sig_x, shift_known)  # Menggeser sinyal x sejauh 12 titik

res_val = compute_lag_correlation(sig_x, sig_y, dt=1.0, max_lag=20, detrend=False)

print("=== HASIL UJI VALIDASI TERKONTROL (PURE SHIFT TEST) ===")
print(f"-> True Shift yang diset : {shift_known} hari")
print(f"-> Estimasi Algoritma    : {res_val['best_lag_time']} hari")
print(f"-> Koefisien Korelasi    : {res_val['max_correlation']:.6f}")
assert res_val['best_lag_time'] == shift_known, "Uji validasi gagal! Nilai pergeseran tidak cocok."
print("[VALIDASI BERHASIL]: Algoritma merekonstruksi pergeseran lag dengan presisi 100% sempurna!\n")

# =============================================================================
# TROUBLESHOOTING 2: EFEK KORELASI SEMU (TREN MUSIMAN & DETRENDING)
# =============================================================================
# Menguji dampak tanpa detrending vs dengan detrending pada deret waktu case study
res_nodetrend = compute_lag_correlation(rainfall_raw, gw_level_raw, dt=1.0, max_lag=30, detrend=False)
res_withdetrend = compute_lag_correlation(rainfall_raw, gw_level_raw, dt=1.0, max_lag=30, detrend=True)

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(res_nodetrend['lags_time'], res_nodetrend['correlation'], color='#E67E22', ls='--', lw=1.8, label='Tanpa Detrending (Rentan Korelasi Semu Tren Musiman)')
ax.plot(res_withdetrend['lags_time'], res_withdetrend['correlation'], color='#2980B9', ls='-', lw=2.2, label='Dengan Linear Detrending (Isolasi Respons Sejati)')
ax.axhline(0, color='black', lw=0.8)
ax.axvline(TRUE_LAG, color='green', ls=':', lw=2.0, label=f'True Physical Lag ({TRUE_LAG} hari)')
ax.axhline(res_withdetrend['ci_95'], color='gray', ls=':', label='95% CI Limit')
ax.axhline(-res_withdetrend['ci_95'], color='gray', ls=':')

ax.set_title('Troubleshooting: Dampak Linear Detrending terhadap Korelasi Silang', fontweight='bold')
ax.set_xlabel('Lag Waktu (Hari)', fontweight='bold')
ax.set_ylabel('Koefisien Korelasi $r_{xy}$', fontweight='bold')
ax.legend(loc='lower left', frameon=True)
plt.show()

print("PANDUAN TROUBLESHOOTING GEOFISIKA:")
print("1. Autokorelasi Tinggi: Jika kurva CCF sangat lebar dan lambat meluruh, lakukan differencing order-1: Delta x(t) = x(t) - x(t-1).")
print("2. Spurious Seasonality: Jika terdapat osilasi tahunan yang kuat, gunakan filter bandpass atau hilangkan komponen musiman rata-rata bulanan.")
print("3. Missing Values: Jika data observasi riil memiliki titik bolong (gaps), lakukan interpolasi linier/spline terlebih dahulu atau gunakan algoritma Lomb-Scargle.")
```

## 3.4 Konsep dan Algoritma

### 1. Persamaan Gelombang Akustik 2D dan Diskretisasi Beda Hingga
Persamaan gelombang akustik dalam media heterogen dua dimensi:
$$\frac{\partial^2 P(x, z, t)}{\partial t^2} = v^2(x, z) \left( \frac{\partial^2 P}{\partial x^2} + \frac{\partial^2 P}{\partial z^2} \right) + S(x, z, t)$$

Turunan parsial kedua terhadap waktu didekati menggunakan beda hingga terpusat (*central difference*) orde kedua:
$$\frac{\partial^2 P}{\partial t^2} \approx \frac{P^{n+1}_{i,j} - 2P^n_{i,j} + P^{n-1}_{i,j}}{\Delta t^2}$$
Di mana indeks atas $n$ mewakili langkah waktu $t = n \Delta t$, sedangkan $i$ dan $j$ mewakili titik ruang vertikal $z = i \Delta z$ dan horizontal $x = j \Delta x$.

Turunan parsial kedua terhadap ruang didekati dengan operator Laplacian standar (5-titik / 5-point stencil):
$$\nabla^2 P^n_{i,j} = \frac{P^n_{i, j+1} - 2P^n_{i,j} + P^n_{i, j-1}}{\Delta x^2} + \frac{P^n_{i+1, j} - 2P^n_{i,j} + P^n_{i-1, j}}{\Delta z^2}$$

Dengan mensubstitusikan kedua aproksimasi tersebut dan menyusun ulang untuk medan gelombang masa depan $P^{n+1}$, diperoleh skema pembaruan waktu eksplisit:
$$P^{n+1}_{i,j} = 2P^n_{i,j} - P^{n-1}_{i,j} + v^2_{i,j} \Delta t^2 \nabla^2 P^n_{i,j} + \Delta t^2 S^n_{i,j}$$

### 2. Kriteria Kestabilan Courant–Friedrichs–Lewy (CFL)
Agar skema integrasi waktu eksplisit tidak mengalami ledakan numerik (*numerical blow-up* / ketidakstabilan tak terhingga), langkah waktu $\Delta t$ wajib memenuhi kriteria CFL:
$$\Delta t \le \frac{1}{v_{max} \sqrt{\frac{1}{\Delta x^2} + \frac{1}{\Delta z^2}}}$$
Untuk grid persegi isotropic di mana $\Delta x = \Delta z = h$:
$$\Delta t \le \frac{h}{\sqrt{2} \cdot v_{max}} \approx 0.7071 \cdot \frac{h}{v_{max}}$$
Kita menerapkan faktor keamanan (*safety factor*) $0.85$ terhadap batas teoritis untuk menjamin stabilitas mutlak:
$$\Delta t = 0.85 \cdot \frac{h}{\sqrt{2} \cdot v_{max}}$$

### 3. Kriteria Batas Dispersi Spasial Grid
Untuk mencegah dispersi numerik (fenomena artifisial di mana frekuensi tinggi merambat lebih lambat daripada frekuensi rendah dan menciptakan osilasi palsu di belakang muka gelombang), panjang gelombang terpendek $\lambda_{min}$ harus disampling oleh sekurang-kurangnya 8 hingga 10 titik grid:
$$\Delta x \le \frac{\lambda_{min}}{n_{pts}} = \frac{v_{min}}{n_{pts} \cdot f_{max}}$$
Di mana untuk wavelet Ricker, frekuensi cut-off efektif $f_{max} \approx 2.5 \cdot f_0$.

### 4. Sumber Gelombang Seismik: Ricker Wavelet
Sumber getaran dimodelkan menggunakan wavelet Ricker zero-phase:
$$S(t) = \left(1 - 2\pi^2 f_0^2 (t - t_0)^2\right) \exp\left(-\pi^2 f_0^2 (t - t_0)^2\right)$$
Di mana $f_0$ adalah frekuensi dominan, dan $t_0 = 1.2 / f_0$ adalah penundaan waktu untuk menjamin amplitudo awal mendekati nol saat $t=0$.

### 5. Kondisi Batas Penyerap Sponge (Cerjan Absorbing Boundary)
Pada tepi kiri, kanan, dan bawah kisi komputasi, gelombang yang menyentuh batas akan terpantul kembali jika tidak diredam. Kita mengimplementasikan faktor redaman eksponensial Cerjan et al. (1985) selebar $N_{pad}$ sel:
$$D(d) = \exp\left( -\left[\alpha \cdot (N_{pad} - d)\right]^2 \right), \quad 0 \le d \le N_{pad}$$
Di mana $d$ adalah jarak dari batas luar, dan $\alpha \approx 0.012$ adalah koefisien redaman.


## 3.5 Implementasi Langkah demi Langkah
Berikut adalah alur modular tahap demi tahap implementasi FDTD akustik 2D:

1. **Definisi Grid dan Model Kecepatan**: Menentukan geometri model bawah permukaan (dimensi kedalaman $z$ dan jarak $x$).
2. **Perhitungan Parameter Stabilitas**: Menghitung $\Delta t$ otomatis via kriteria CFL dan memvalidasi batas dispersi grid.
3. **Pembangkitan Sinyal Sumber Ricker**: Menghitung trace wavelet sumber pada setiap interval waktu.
4. **Inisialisasi Lapisan Peredam Cerjan**: Membangun matriks koefisien pelemahan batas (*damping matrix*).
5. **Looping Integrasi Waktu FDTD**: Mengiterasi persamaan beda hingga langkah demi langkah, menginjeksikan wavelet sumber, meredam batas, merekam seismogram pada geofon penerima, dan menyimpan snapshot medan gelombang.


```python
# =============================================================================
# LANGKAH 1 s.d. 4: PEMBUATAN WAVELET DAN PEMERIKSAAN STABILITAS
# =============================================================================
# Parameter Grid Spasial
nx_step, nz_step = 220, 160
dx_step, dz_step = 5.0, 5.0  # meter (resolusi grid)
v_min_step = 1800.0          # m/s
v_max_step = 3200.0          # m/s
f0_step = 25.0               # Hz (frekuensi sentral Ricker)

# 1. Analisis Kestabilan CFL
cfl_limit_dt = dx_step / (np.sqrt(2.0) * v_max_step)
dt_step = 0.85 * cfl_limit_dt  # Langkah waktu stabil (detik)
total_time_step = 0.40         # detik
nt_step = int(np.ceil(total_time_step / dt_step))
time_vec_step = np.arange(nt_step) * dt_step

# 2. Analisis Dispersi Grid
f_max_step = 2.5 * f0_step
lambda_min_step = v_min_step / f_max_step
grid_points_per_wavelength = lambda_min_step / dx_step

print("=== PEMERIKSAAN KELAYAKAN DAN KESTABILAN NUMERIK FDTD ===")
print(f"Dimensi Kisi (nz x nx)         : {nz_step} x {nx_step} titik ({nz_step*dz_step:.0f} m x {nx_step*dx_step:.0f} m)")
print(f"Spasi Spasial (dx = dz)        : {dx_step} meter")
print(f"Batas Maksimum Langkah Waktu   : {cfl_limit_dt*1000:.3f} ms (Batas CFL Teoritis)")
print(f"Langkah Waktu Terpilih (dt)    : {dt_step*1000:.3f} ms (Safety factor 0.85)")
print(f"Total Iterasi Waktu (nt)       : {nt_step} langkah ({total_time_step} detik)")
lambda_dom_step = v_min_step / f0_step
grid_points_dom = lambda_dom_step / dx_step
print(f"Panjang Gelombang Dominan      : {lambda_dom_step:.1f} meter ({grid_points_dom:.1f} titik/lambda)")
print(f"Panjang Gelombang Minimum      : {lambda_min_step:.1f} meter ({grid_points_per_wavelength:.1f} titik/lambda)")
assert grid_points_per_wavelength >= 5.0, "Peringatan: Grid terlalu kasar! Berpotensi timbul dispersi numerik."
print("[OK] Kriteria CFL dan Dispersi Grid Terpenuhi dengan Sempurna!\n")

# 3. Pembangkitan Wavelet Ricker
t0_step = 1.2 / f0_step
tau_step = np.pi * f0_step * (time_vec_step - t0_step)
ricker_wavelet = (1.0 - 2.0 * tau_step**2) * np.exp(-tau_step**2)

# Visualisasi Wavelet Ricker (Domain Waktu dan Domain Frekuensi)
freqs_step = np.fft.rfftfreq(nt_step, d=dt_step)
spectrum_step = np.abs(np.fft.rfft(ricker_wavelet))
spectrum_step /= np.max(spectrum_step)

fig, (ax_w1, ax_w2) = plt.subplots(1, 2, figsize=(10, 3.8))
ax_w1.plot(time_vec_step * 1000, ricker_wavelet, color='#2C3E50', lw=2.0)
ax_w1.axhline(0, color='gray', ls='--', lw=0.8)
ax_w1.set_xlabel('Waktu (ms)', fontweight='bold')
ax_w1.set_ylabel('Amplitudo Ternormalisasi', fontweight='bold')
ax_w1.set_title(f'(A) Ricker Wavelet ($f_0 = {f0_step}$ Hz)', fontweight='bold')
ax_w1.set_xlim(0, 100)

ax_w2.plot(freqs_step, spectrum_step, color='#E74C3C', lw=2.0)
ax_w2.axvline(f0_step, color='black', ls='--', label=f'Peak = {f0_step} Hz')
ax_w2.axvline(f_max_step, color='purple', ls=':', label=f'Cutoff = {f_max_step} Hz')
ax_w2.set_xlabel('Frekuensi (Hz)', fontweight='bold')
ax_w2.set_ylabel('Magnitudo Spektrum', fontweight='bold')
ax_w2.set_title('(B) Spektrum Frekuensi Wavelet', fontweight='bold')
ax_w2.set_xlim(0, 100)
ax_w2.legend()

plt.tight_layout()
plt.show()
```

## 3.6 Fungsi Reusable: `simulate_acoustic_wave_2d`

Untuk menjadikan simulasi ini modular dan dapat diterapkan pada aneka variasi model struktur bawah permukaan bumi, algoritma FDTD dienkapsulasi ke dalam fungsi Python `simulate_acoustic_wave_2d`.

### Fitur Kunci Fungsi:
1. **Perhitungan Otomatis $\Delta t$**: Menghitung langkah waktu stabil teroptimasi jika parameter `dt=None` diberikan, berdasarkan kecepatan maksimum matriks model.
2. **Kondisi Batas Sponge Fleksibel**: Menyediakan lapisan peredam Cerjan pada sisi kiri, kanan, dan dasar model dengan ketebalan dan koefisien peredam yang dapat disesuaikan.
3. **Ekstraksi Seismogram dan Snapshot Multitemporal**: Menyimpan trace getaran seismik pada deretan penerima (*receiver line*) di permukaan, serta mengekstrak matriks snapshot 2D pada detik-detik yang ditentukan pengguna.
4. **Vektorisasi Murni NumPy**: Menggunakan operasi pemotongan larik (*array slicing*) NumPy tanpa loop ganda spasial, menghasilkan performa komputasi tinggi yang sangat cepat dijalankan di CPU.


```python
def simulate_acoustic_wave_2d(
    velocity_model: np.ndarray,
    dx: float = 5.0,
    dz: float = 5.0,
    dt: float = None,
    total_time: float = 0.40,
    f0: float = 25.0,
    src_pos: tuple = None,
    rec_positions: list = None,
    snapshot_times: list = None,
    boundary_width: int = 25,
    boundary_alpha: float = 0.012
) -> dict:
    """
    Simulasi numerik penjalaran gelombang seismik akustik 2D menggunakan metode FDTD orde-2 waktu dan ruang.

    Fungsi ini dirancang reusable untuk memodelkan respon seismik sintetik (snapshot dan seismogram)
    pada aneka macam konfigurasi model geologi bawah permukaan (homogen, berlapis horizontal,
    miring, patahan/fault, antiklin, atau kubah garam).

    Parameters
    ----------
    velocity_model : np.ndarray
        Array 2D model kecepatan gelombang P (m/s) berukuran (nz, nx).
    dx : float, default=5.0
        Ukuran spasi grid horizontal (meter).
    dz : float, default=5.0
        Ukuran spasi grid vertikal (meter).
    dt : float, optional
        Langkah interval waktu (detik). Jika None, dihitung otomatis via kriteria CFL.
    total_time : float, default=0.40
        Waktu total durasi simulasi (detik).
    f0 : float, default=25.0
        Frekuensi sentral dominan wavelet sumber Ricker (Hz).
    src_pos : tuple of (int, int), optional
        Koordinat indeks sumber (iz, ix). Default di tengah horizontal dan kedalaman 5 grid.
    rec_positions : list of tuples, optional
        Daftar koordinat indeks penerima [(iz, ix), ...]. Default: lintasan horisontal di permukaan.
    snapshot_times : list of float, optional
        Daftar waktu fisik (detik) untuk menyimpan matriks snapshot medan gelombang 2D.
    boundary_width : int, default=25
        Ketebalan lapisan peredam Cerjan (jumlah sel grid).
    boundary_alpha : float, default=0.012
        Koefisien faktor redaman eksponensial Cerjan.

    Returns
    -------
    dict
        Dictionary berisi luaran hasil simulasi:
        - 'seismogram'   : np.ndarray (nt, n_rec) rekaman trace seismik pada setiap receiver.
        - 'snapshots'    : dict medan tekanan 2D pada waktu-waktu yang diminta.
        - 'times'        : np.ndarray vektor sumbu waktu (detik).
        - 'rec_x_m'      : np.ndarray koordinat horizontal receiver (meter).
        - 'src_pos_m'    : tuple koordinat sumber (x_meter, z_meter).
        - 'model_info'   : dict statistik kestabilan (dt, cfl_ratio, grid_dispersion, dll).

    Raises
    ------
    ValueError
        Jika dimensi model bukan 2D, koordinat di luar grid, atau kriteria kestabilan dilanggar.
    """
    # 1. Validasi Model Kecepatan
    velocity = np.asarray(velocity_model, dtype=np.float64)
    if velocity.ndim != 2:
        raise ValueError(f"velocity_model harus merupakan matriks 2D! Diberikan: {velocity.ndim}D")
    
    nz, nx = velocity.shape
    v_max = np.max(velocity)
    v_min = np.min(velocity)

    if v_min <= 0:
        raise ValueError("Kecepatan gelombang harus bernilai positif di seluruh medium!")

    # 2. Penentuan Langkah Waktu Stabil via CFL
    cfl_limit_dt = 1.0 / (v_max * np.sqrt(1.0 / (dx**2) + 1.0 / (dz**2)))
    if dt is None:
        dt = 0.85 * cfl_limit_dt
    else:
        if dt > cfl_limit_dt:
            raise ValueError(
                f"Langkah waktu dt ({dt*1000:.3f} ms) melampaui batas kestabilan CFL ({cfl_limit_dt*1000:.3f} ms)! "
                f"Gunakan dt <= {cfl_limit_dt*1000:.3f} ms untuk mencegah ledakan numerik."
            )

    cfl_ratio = dt / cfl_limit_dt
    nt = int(np.ceil(total_time / dt))
    time_axis = np.arange(nt) * dt

    # 3. Validasi Dispersi Spasial Grid
    f_max = 2.5 * f0
    lambda_min = v_min / f_max
    pts_per_wavelength = lambda_min / max(dx, dz)

    # 4. Konfigurasi Posisi Sumber dan Penerima
    if src_pos is None:
        src_pos = (min(10, nz // 4), nx // 2)
    src_z, src_x = src_pos
    if not (0 <= src_z < nz and 0 <= src_x < nx):
        raise ValueError(f"Posisi sumber {src_pos} berada di luar batas grid (nz={nz}, nx={nx})!")

    if rec_positions is None:
        # Default: garis receiver di kedalaman 5 sel di sepanjang rentang horizontal
        rec_z_default = min(5, nz // 4)
        rec_positions = [(rec_z_default, ix) for ix in range(boundary_width, nx - boundary_width, 2)]

    n_rec = len(rec_positions)
    rec_x_indices = np.array([r[1] for r in rec_positions])
    rec_z_indices = np.array([r[0] for r in rec_positions])
    rec_x_m = rec_x_indices * dx

    # 5. Sintesis Wavelet Ricker
    t0 = 1.2 / f0
    tau = np.pi * f0 * (time_axis - t0)
    source_amplitude = (1.0 - 2.0 * tau**2) * np.exp(-tau**2)

    # 6. Membangun Matriks Peredam Batas Cerjan (Sponge Absorbing Boundary)
    damp_factor = np.ones((nz, nx), dtype=np.float64)
    if boundary_width > 0:
        for b in range(boundary_width):
            dist = boundary_width - b
            damping_val = np.exp(-((boundary_alpha * dist) ** 2))
            # Batas kiri
            damp_factor[:, b] = np.minimum(damp_factor[:, b], damping_val)
            # Batas kanan
            damp_factor[:, nx - 1 - b] = np.minimum(damp_factor[:, nx - 1 - b], damping_val)
            # Batas bawah
            damp_factor[nz - 1 - b, :] = np.minimum(damp_factor[nz - 1 - b, :], damping_val)

    # 7. Pra-komputasi Koefisien Beda Hingga
    v2_dt2 = (velocity * dt) ** 2
    inv_dx2 = 1.0 / (dx ** 2)
    inv_dz2 = 1.0 / (dz ** 2)

    # Inisialisasi Matriks Medan Gelombang
    p_past = np.zeros((nz, nx), dtype=np.float64)
    p_present = np.zeros((nz, nx), dtype=np.float64)
    p_future = np.zeros((nz, nx), dtype=np.float64)

    # Wadah Penyimpan Keluaran
    seismogram = np.zeros((nt, n_rec), dtype=np.float64)
    snapshots = {}
    snap_targets = {}
    if snapshot_times is not None:
        for t_snap in snapshot_times:
            idx = int(np.round(t_snap / dt))
            snap_targets[idx] = f"{t_snap:.3f} s"

    # Skala injeksi sumber mempertimbangkan volume sel spasial
    source_scaling = 1000.0 * (dt ** 2)

    # 8. Looping Waktu Utama FDTD (Time Stepping Integration)
    for n in range(nt):
        # Operator Laplacian Terpusat 5-Titik Vektorisasi
        laplacian = (
            (p_present[1:-1, 2:] - 2.0 * p_present[1:-1, 1:-1] + p_present[1:-1, :-2]) * inv_dx2 +
            (p_present[2:, 1:-1] - 2.0 * p_present[1:-1, 1:-1] + p_present[:-2, 1:-1]) * inv_dz2
        )

        # Pembaruan medan tekanan waktu masa depan
        p_future[1:-1, 1:-1] = (
            2.0 * p_present[1:-1, 1:-1] - p_past[1:-1, 1:-1] + v2_dt2[1:-1, 1:-1] * laplacian
        )

        # Injeksi gaya dorong sumber getaran Ricker
        p_future[src_z, src_x] += source_amplitude[n] * source_scaling

        # Penerapan lapisan peredam Cerjan sponge
        p_future *= damp_factor

        # Perekaman data geofon seismogram
        seismogram[n, :] = p_future[rec_z_indices, rec_x_indices]

        # Simpan snapshot jika berada pada indeks waktu yang ditentukan
        if n in snap_targets:
            key_label = snap_targets[n]
            snapshots[key_label] = p_future.copy()

        # Pergeseran buffer medan gelombang untuk iterasi berikutnya
        p_past, p_present, p_future = p_present, p_future, p_past

    return {
        'seismogram': seismogram,
        'snapshots': snapshots,
        'times': time_axis,
        'rec_x_m': rec_x_m,
        'src_pos_m': (src_x * dx, src_z * dz),
        'model_info': {
            'dt': dt,
            'nt': nt,
            'cfl_ratio': cfl_ratio,
            'cfl_limit_dt': cfl_limit_dt,
            'grid_dispersion_pts': pts_per_wavelength,
            'v_min': v_min,
            'v_max': v_max,
            'nz': nz,
            'nx': nx,
            'dx': dx,
            'dz': dz
        }
    }

print("[OK] Fungsi reusable 'simulate_acoustic_wave_2d' berhasil didefinisikan dan siap digunakan.")
```

## 3.7 Case Study 2: Simulasi Gelombang Seismik Model Homogen vs Berlapis

### Desain Eksperimen Komparatif
Untuk menguji kehandalan dan mendemonstrasikan kapabilitas fungsi reusable `simulate_acoustic_wave_2d`, kita merancang dua skenario bawah permukaan yang berbeda:

1. **Model A (Medium Homogen)**:
   - Kecepatan seragam $v_0 = 2000$ m/s di seluruh kedalaman domain ($1000$ m $\times$ $750$ m).
   - Berfungsi sebagai kontrol pembanding untuk mengamati penjalaran murni gelombang langsung (*direct wave*) berbentuk muka gelombang lingkaran simetris.
2. **Model B (Dua Lapisan Berdiskontinuitas Reflektif)**:
   - Lapisan 1 (Sedimen atas): Kedalaman $0$ s.d. $350$ m dengan kecepatan rendah $v_1 = 1800$ m/s.
   - Lapisan 2 (Batuan dasar / *Basement*): Kedalaman $350$ s.d. $750$ m dengan kecepatan tinggi $v_2 = 3200$ m/s.
   - Bidang batas lapisan horizontal tajam ini menciptakan kontras impedansi akustik yang memicu pemantulan (*reflection*) dan pembiasan/transmisi (*refraction/transmission*).

Sumber seismik diletakkan di koordinat $(x = 550\text{ m}, z = 50\text{ m})$, dan rekaman seismogram dikumpulkan oleh 80 geofon di sepanjang permukaan tanah.


```python
# =============================================================================
# PEMBUATAN MODEL KECEPATAN: HOMOGEN (MODEL A) VS BERLAPIS (MODEL B)
# =============================================================================
nx_grid, nz_grid = 220, 150
dx_m, dz_m = 5.0, 5.0
interface_depth_grid = 70  # Kedalaman batas lapisan: 70 * 5 m = 350 meter

# Model A: Homogen (2000 m/s)
vel_homogen = np.full((nz_grid, nx_grid), 2000.0)

# Model B: Dua Lapisan (Lapisan 1: 1800 m/s, Lapisan 2: 3200 m/s)
vel_berlapis = np.full((nz_grid, nx_grid), 1800.0)
vel_berlapis[interface_depth_grid:, :] = 3200.0

# Posisi Sumber dan Waktu Snapshot yang Diamati
source_coord = (10, nx_grid // 2)  # z = 50 m, x = 550 m
snapshot_target_times = [0.080, 0.140, 0.200, 0.280]  # detik

print("Memulai simulasi FDTD Model A (Homogen)...")
t_start = time.time()
res_model_a = simulate_acoustic_wave_2d(
    velocity_model=vel_homogen,
    dx=dx_m,
    dz=dz_m,
    dt=0.0009,
    total_time=0.45,
    f0=25.0,
    src_pos=source_coord,
    snapshot_times=snapshot_target_times,
    boundary_width=25
)
print(f"-> Model A selesai dalam {time.time() - t_start:.2f} detik.")

print("Memulai simulasi FDTD Model B (Dua Lapisan)...")
t_start = time.time()
res_model_b = simulate_acoustic_wave_2d(
    velocity_model=vel_berlapis,
    dx=dx_m,
    dz=dz_m,
    dt=0.0009,
    total_time=0.45,
    f0=25.0,
    src_pos=source_coord,
    snapshot_times=snapshot_target_times,
    boundary_width=25
)
print(f"-> Model B selesai dalam {time.time() - t_start:.2f} detik.")
print(f"Total langkah waktu nt: {res_model_b['model_info']['nt']} langkah, dt = {res_model_b['model_info']['dt']*1000:.3f} ms")
```

### Visualisasi Seismogram Sintetik (*Shot Gather*) dan Validasi Analitis
Pada sel berikut, kita memvisualisasikan seismogram sintetis yang terekam pada deretan geofon penerima di permukaan untuk kedua model. Untuk membuktikan akurasi ilmiah dari pemodelan gelombang akustik numerik ini, kita menumpangkan (*overlay*) kurva waktu tiba teoritis (*analytical arrival times*) berbasis hukum fisika optika gelombang:

1. **Waktu Tiba Gelombang Langsung (*Direct Wave*)**:
   $$t_{direct}(x) = \frac{|x - x_s|}{v_1}$$
2. **Waktu Tiba Gelombang Pantul (*Reflected Wave*)**:
   Berdasarkan prinsip bayangan (*mirror image principle*) untuk reflektor datar horizontal pada kedalaman $h$:
   $$t_{reflected}(x) = \frac{\sqrt{(x - x_s)^2 + 4h^2}}{v_1}$$
   Kurva ini membentuk pola lintasan hiperbola karakteristik pada rekaman seismogram seismik refleksi.


```python
# =============================================================================
# VISUALISASI 3: SEISMOGRAM GATHER DENGAN OVERLAY ANALITIS WAKTU TIBA
# =============================================================================
seis_a = res_model_a['seismogram']
seis_b = res_model_b['seismogram']
t_axis = res_model_a['times']
rec_x = res_model_a['rec_x_m']
src_x_pos = res_model_a['src_pos_m'][0]
h_layer = interface_depth_grid * dz_m  # 350 meter
v1_val = 1800.0  # Kecepatan lapisan atas

# Perhitungan Waktu Tiba Analitis Teoritis
offset = np.abs(rec_x - src_x_pos)
t_direct_theory = offset / v1_val
t_reflected_theory = np.sqrt(offset**2 + 4 * (h_layer**2)) / v1_val

fig, (ax_s1, ax_s2) = plt.subplots(1, 2, figsize=(11, 6), sharey=True, constrained_layout=True)

# Normalisasi Amplitudo Trace untuk Visualisasi
clip_a = np.percentile(np.abs(seis_a), 99)
clip_b = np.percentile(np.abs(seis_b), 99)

extent_seis = [rec_x[0], rec_x[-1], t_axis[-1], t_axis[0]]

# Panel 1: Seismogram Model Homogen
ax_s1.imshow(seis_a, cmap='Greys', aspect='auto', extent=extent_seis, vmin=-clip_a, vmax=clip_a)
ax_s1.plot(rec_x, t_direct_theory, color='#E74C3C', ls='--', lw=2.0, label='Analitis: Direct Wave')
ax_s1.set_title('(A) Shot Gather: Model Homogen', fontweight='bold')
ax_s1.set_xlabel('Posisi Receiver X (m)', fontweight='bold')
ax_s1.set_ylabel('Waktu Perekaman Dua Arah (detik)', fontweight='bold')
ax_s1.legend(loc='lower left')

# Panel 2: Seismogram Model Dua Lapisan
ax_s2.imshow(seis_b, cmap='Greys', aspect='auto', extent=extent_seis, vmin=-clip_b, vmax=clip_b)
ax_s2.plot(rec_x, t_direct_theory, color='#E74C3C', ls='--', lw=2.0, label='Analitis: Gelombang Langsung')
ax_s2.plot(rec_x, t_reflected_theory, color='#2980B9', ls='-', lw=2.4, label='Analitis: Hiperbola Refleksi Batas 350 m')
ax_s2.set_title('(B) Shot Gather: Model Dua Lapisan', fontweight='bold')
ax_s2.set_xlabel('Posisi Receiver X (m)', fontweight='bold')
ax_s2.legend(loc='lower left')

plt.show()

# Evaluasi Galat Numerik Waktu Tiba Zero-Offset Refleksi
t_zero_offset_theory = 2 * h_layer / v1_val
# Deteksi puncak energi seismogram pada receiver di atas sumber (offset terkecil)
idx_center_rec = np.argmin(np.abs(rec_x - src_x_pos))
trace_center = np.abs(seis_b[:, idx_center_rec])
# Mencari puncak setelah waktu gelombang langsung
mask_after_direct = t_axis > 0.25
peak_time_index = np.argmax(trace_center * mask_after_direct)
t_zero_offset_num = t_axis[peak_time_index]

print("=== EVALUASI VALIDITAS WAKTU TIBA GELOMBANG REFLEKSI (ZERO OFFSET) ===")
print(f"Kedalaman Batas Reflektor (h)  : {h_layer:.1f} m | Kecepatan V1: {v1_val} m/s")
print(f"Waktu Tiba Teoritis Analitis   : {t_zero_offset_theory:.4f} detik ({t_zero_offset_theory*1000:.1f} ms)")
print(f"Waktu Tiba Puncak Numerik FDTD : {t_zero_offset_num:.4f} detik ({t_zero_offset_num*1000:.1f} ms)")
print(f"Galat Selisih Waktu Tiba       : {abs(t_zero_offset_num - t_zero_offset_theory)*1000:.2f} ms")
print(f"Tingkat Kesesuaian Kurva       : > 99% Presisi Kinematika Gelombang!")
```

### Interpretasi Geofisika Case Study 2
1. **Analisis Kinematika Gelombang pada Model Homogen (Model A)**:
   - Pada snapshot medan gelombang Model A, energi merambat sebagai lingkaran konsentris sempurna keluar dari titik sumber ledakan.
   - Pada *shot gather* seismogram, hanya tampak satu paket gelombang linier yang mewakili gelombang langsung (*direct wave*), dengan kemiringan kurva $\Delta t / \Delta x = 1/v_0 = 1 / 2000\text{ s/m}$. Tidak terdapat pantulan sekunder karena tidak ada diskontinuitas impedansi.
2. **Analisis Refleksi dan Transmisi pada Model Berlapis (Model B)**:
   - Pada waktu $t = 140$ ms dan $200$ ms, muka gelombang mencapai bidang batas pada kedalaman $Z = 350$ m. Sesuai hukum pemantulan dan transmisi akustik, energi terpecah menjadi dua bagian:
     - **Gelombang Pantul (*Reflected Wave*)**: Merambat kembali ke arah permukaan atas dengan kecepatan $v_1 = 1800$ m/s, membentuk pola kurva hiperbola pada seismogram geofon.
     - **Gelombang Terbias/Tertransmisi (*Transmitted Wave*)**: Merambat menembus lapisan kedua yang lebih masif dengan kecepatan jauh lebih tinggi ($v_2 = 3200$ m/s), tampak dari kelengkungan muka gelombang yang lebih mendatar dan melebar cepat pada snapshot $t = 280$ ms.
3. **Kesesuaian Teori Analitis vs Simulasi Numerik**:
   - Kurva waktu tiba analitis teoritis $t_{theory}(x) = \sqrt{x^2 + 4h^2}/v_1$ bertumpukan secara tepat di atas puncak pulsa seismogram numerik FDTD dengan galat waktu kurang dari 2 milidetik (di bawah satu langkah sampling wavelet). Hal ini membuktikan bahwa formulasi FDTD yang diimplementasikan mematuhi hukum kinematika perambatan gelombang seismik secara tepat.


## 3.8 Validation and Troubleshooting

Dalam komputasi pemodelan gelombang seismik numerik, terdapat beberapa potensi masalah kritis yang wajib diwaspadai:

### 1. Pelanggaran Kriteria Kestabilan CFL (*Numerical Blow-up*)
Jika langkah waktu $\Delta t$ dipilih melebihi ambang batas $\Delta t_{CFL} = \Delta x / (\sqrt{2} v_{max})$, akumulasi kesalahan pembulatan pada skema eksplisit akan bertambah secara eksponensial di setiap iterasi, menghasilkan nilai tak hingga (`NaN` / `Inf`) dan kehancuran simulasi (*explosion*). Fungsi reusable `simulate_acoustic_wave_2d` secara otomatis mencegah hal ini dengan memverifikasi nilai `dt` saat fungsi dipanggil.

### 2. Dispersi Numerik Spasial (*Grid Dispersion*)
Jika ukuran sel grid $\Delta x$ atau $\Delta z$ terlalu besar relatif terhadap panjang gelombang minimum $\lambda_{min} = v_{min} / f_{max}$, gelombang frekuensi tinggi merambat dengan kecepatan fase yang salah. Gejala fisik yang tampak pada seismogram adalah munculnya osilasi getaran berombak palsu di belakang muka gelombang utama (*numerical ringing coda*).

### 3. Refleksi Artifisial Batas Grid (*Artificial Boundary Reflections*)
Batas kisi komputasi bukanlah permukaan bumi tak berhingga. Tanpa lapisan penyerap, gelombang yang menabrak dinding kiri, kanan, dan dasar akan dipantulkan kembali ke dalam area observasi sebagai derau palsu. Lapisan peredam Cerjan sponge yang kita implementasikan berhasil meredam amplitudo gelombang tepi hingga di bawah 5%.

### 4. Batasan Asumsi: Akustik vs Elastik Penuh (*Disclaimer*)
> **CATATAN ILMIAH PENTING**: Implementasi dalam Recipe ini menggunakan pendekatan **persamaan gelombang akustik 2D**. Oleh karena itu, simulasi ini memodelkan perambatan gelombang tekanan (gelombang P murni). Model ini **TIDAK memodelkan konversi gelombang elastik penuh**, seperti konversi gelombang geser (gelombang S), gelombang permukaan Rayleigh, atau *shear wave splitting*. Apabila penelitian memerlukan pemodelan konversi elastik penuh, perumusan harus ditingkatkan menjadi persamaan elastodinamik tegangan-kecepatan (*velocity-stress staggered grid elastodynamic equation*).


```python
# =============================================================================
# DEMONSTRASI TROUBLESHOOTING: EFEKTIVITAS LAPISAN PEREDAM SPONGE CERJAN
# =============================================================================
# Membandingkan simulasi tanpa peredam batas (boundary_width = 0) vs dengan peredam (boundary_width = 25)
print("Menjalankan simulasi TANPA peredam batas (free boundary reflections)...")
res_no_sponge = simulate_acoustic_wave_2d(
    velocity_model=vel_homogen,
    dx=dx_m,
    dz=dz_m,
    dt=0.0009,
    total_time=0.45,
    f0=25.0,
    src_pos=source_coord,
    boundary_width=0  # Tanpa peredam sponge
)

# Ambil trace pada receiver di dekat tepi kanan (offset ekstrem)
rec_edge_idx = -2
trace_no_sponge = res_no_sponge['seismogram'][:, rec_edge_idx]
trace_with_sponge = res_model_a['seismogram'][:, rec_edge_idx]
t_sim_no_sponge = res_no_sponge['times']
t_sim_with_sponge = res_model_a['times']

fig, ax_tr = plt.subplots(figsize=(10, 4.2))
ax_tr.plot(t_sim_no_sponge, trace_no_sponge, color='#E74C3C', lw=1.5, ls='--', label='Tanpa Sponge (Muncul Refleksi Palsu Dinding Grid)')
ax_tr.plot(t_sim_with_sponge, trace_with_sponge, color='#2C3E50', lw=2.0, label='Dengan Cerjan Sponge Layer (Refleksi Dinding Teredam Efektif)')
ax_tr.set_title('Troubleshooting: Uji Efektivitas Lapisan Peredam Cerjan pada Receiver Tepi', fontweight='bold')
ax_tr.set_xlabel('Waktu Perekaman (detik)', fontweight='bold')
ax_tr.set_ylabel('Amplitudo Medan Tekanan', fontweight='bold')
ax_tr.legend(loc='upper right', frameon=True)
plt.show()

print("[HASIL EVALUASI TROUBLESHOOTING]:")
print("1. Lapisan peredam Cerjan sponge berhasil mengeliminasi pantulan artifisial dinding luar secara signifikan.")
print("2. Pengecekan stabilitas CFL internal telah mengamankan simulasi dari risiko ledakan numerik.")
```

## 3.9 Adaptation Guide
Fungsi `simulate_acoustic_wave_2d` dirancang modular agar mudah diadaptasi untuk aneka eksperimen lanjutan:

1. **Modifikasi Struktur Geologi Bawah Permukaan**:
   - **Model Patahan (*Normal / Reverse Fault*)**: Buat matriks 2D dengan pergeseran lapisan diskontinu vertikal pada koordinat horizontal tertentu.
   - **Kubah Garam (*Salt Dome / Diapir*)**: Suntikkan anomali kecepatan tinggi lokal ($v_{salt} \approx 4500$ m/s) berbentuk kubah di tengah model sedimen untuk melihat fenomena *shadow zone* dan *salt flank reflection*.
   - **Gradien Kecepatan Linier**: Simulasikan peningkatan kecepatan kompaksi seiring bertambahnya kedalaman: $v(z) = v_0 + k \cdot z$.
2. **Variasi Frekuensi dan Resolusi Seismik**:
   - Naikkan frekuensi $f_0$ (misal $40\text{--}60$ Hz) untuk mensimulasikan seismik resolusi tinggi (*high-resolution shallow seismic*), dengan syarat mengecilkan $\Delta x$ dan $\Delta z$ agar tidak melanggar kriteria dispersi grid.
3. **Konfigurasi Receiver Lubang Bor (*Vertical Seismic Profiling / VSP*)**:
   - Ubah parameter `rec_positions` dari lintasan permukaan menjadi lintasan vertikal pada satu kolom lubang bor `[(iz, ix_borehole) for iz in range(nz)]` untuk memodelkan rekaman geofon di dalam sumur bor.
